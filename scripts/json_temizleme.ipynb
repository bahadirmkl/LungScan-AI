```python
import json

# JSON dosyasını yükle
with open('/content/annotations.json', 'r') as file:
    data = json.load(file)

# Görüntüler ve annotasyonlar için filtrelenmiş listeler oluştur
filtered_images = []
filtered_annotations = []

# Tüm görüntüler için annotasyonları kontrol et
for image_info in data['images']:
    # İlgili annotasyonları bul ve segmentasyon alanlarını kontrol et
    annotations_for_image = [anno for anno in data['annotations'] if anno['image_id'] == image_info['id']]
    has_segmentation = False

    for anno in annotations_for_image:
        for segment in anno['segmentation']:
            if segment:  # Segmentasyon alanı boş değilse
                has_segmentation = True
                filtered_annotations.append(anno)  # Bu annotasyonu filtrelenmiş listeye ekle

    # Eğer bu görüntü için en az bir dolu segmentasyon alanı varsa, görüntüyü filtrelenmiş listeye ekle
    if has_segmentation:
        filtered_images.append(image_info)

# Temizlenmiş verileri içeren yeni bir JSON nesnesi oluştur
cleaned_data = {
    "info": data['info'],
    "images": filtered_images,
    "annotations": filtered_annotations,
    "categories": data['categories']
}

# Temizlenmiş verileri yeni bir JSON dosyasına yaz
cleaned_annotations_path = '/content/cleaned_annotations.json'
with open(cleaned_annotations_path, 'w') as file:
    json.dump(cleaned_data, file, indent=4)

print(f"Temizlenmiş annotasyonlar {cleaned_annotations_path} dosyasına kaydedildi.")
```
